```python
import torch
from diffusers import StableDiffusionXLPipeline
```

```python
# specify which stable-diffusion model
model_id = "stabilityai/stable-diffusion-xl-base-1.0"
```

```python
# load model
pipe = StableDiffusionXLPipeline.from_pretrained(
    model_id,
    dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    use_safetensors=True
)
```

```python
# use GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
pipe = pipe.to(device)
```

```python
# create web app and user-interface
import gradio as gr
```

```python
# css style
css = """
button.primary {
    background-color: #4682b4;
    border-color: #c0c0c0;
    color: #ffffff;
}

h1 {
    font-family: 'Georgia', serif;
    color: #4682b4;
}

"""
```

```python
# design room decor as per the selected style and color theme
def design(room_style, room_type, color_scheme):

  # describe things to avoid during generation process
  negative_prompt = "blurry, distorted, flashy bright gaudy colors, cartoon, sketch"

  # generate text description
  prompt = "photorealistic, high quality " + room_style + " style " + room_type + ", using soft color tones and shades of " + color_scheme + " and white"

  image = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=30,
    guidance_scale=7.5,
    strength = 0.75,
    width=512,
    height=512
  ).images[0]

  image.save("output.jpg")

  return image
```

```python
# user-interface

demo = gr.Interface(
    fn=design,
    inputs=[
        gr.Dropdown(
            choices=["Contemporary", "Bohemian", "Modern", "Scandinavian"],
            value="Contemporary",  # Default selected option
            label="Room Style",
        ),
        gr.Dropdown(
            choices=["Living Room", "Bedroom", "Kitchen", "Office", "Bathroom"],
            value="Living Room",
            label="Room Type",
        ),
        gr.Dropdown(
            choices=["Blue", "Peach", "Plum", "Purple", "Turquoise", "Brown", "Grey"],
            value="Blue",
            label="Color Choice",
        )
    ],
    outputs=gr.Image(label="room design", value="sample.jpg"),
    title="Room Decor",
    flagging_mode='never',
    css = css,
)
```

```python
demo.launch(share=True)
```
